```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("sales_clean.csv", dtype={"zip": str})
print(df.shape)
```

```python
for col in ["price", "qty"]:
    print(col, "mean:", round(df[col].mean(), 2),
          "median:", round(df[col].median(), 2),
          "std:", round(df[col].std(), 2))
```

I would report the median ($37.53) for price. The mean ($55.52) is skewed by high-priced items and the median is more resistant to being skewed.

```python
df["price"].hist(bins=20)
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.title("Distribution of Price")
plt.show()
```

The histogram is right skewed.

```python
df["revenue"] = df["price"] * df["qty"]
df.head()
```

```python
df[["price", "qty", "revenue"]].corr()
```

```python
plt.scatter(df["qty"], df["revenue"])
plt.xlabel("Qty")
plt.ylabel("Revenue")
plt.title("Qty vs Revenue")
plt.show()
```

Revenue is driven by price (r ≈ 0.82) more than by quantity (r ≈ 0.31). This correlation doesn't prove causation, and a few very expensive outliers could be swaying the results.
