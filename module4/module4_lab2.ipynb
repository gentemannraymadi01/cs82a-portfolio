```python
import pandas as pd

df = pd.read_csv('allbacks.csv')
df
```

# median cover type

```python
n = len(df)
sorted_covers = sorted(df['cover'])
median_idx = n // 2  # 0-based index of the middle value (n=15 is odd)
median_cover = sorted_covers[median_idx]

print("Sorted covers:", sorted_covers)
print("n =", n, "-> middle position (0-based index) =", median_idx)
print("Median cover type:", median_cover)
```

The median cover type is paperback and there are only 2 possible types so paperback is the most common.

# mean weight of all books

```python
mean_weight_all = df['weight'].mean()
print("Mean weight (all books):", mean_weight_all)
```

The mean is about 706.7g, so on average each book weighs that much.

# mean weight within each cover type

```python
group_means = df.groupby('cover')['weight'].mean()
print(group_means)
```

The mean weight is very different between the two types (about 168g higher for hardcovers), which makes sense since hardcovers use heavier binding materials. For a bookstore this matters for shipping costs and storage weight, so mean weight by book type is a more useful reference than the overall mean.

# verification

```python
hb_weights = df.loc[df['cover'] == 'hb', 'weight'].tolist()
hand_sum = sum(hb_weights)
hand_count = len(hb_weights)
hand_mean = hand_sum / hand_count

print("Hardcover weights:", hb_weights)
print(f"Sum = {hand_sum}, Count = {hand_count}, Mean = {hand_sum} / {hand_count} = {hand_mean:.2f}")
print("Matches pandas groupby result:", round(hand_mean, 2) == round(group_means['hb'], 2))
```
