```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

df = pd.read_csv('truck.csv')
df.head()
```

```python
r, p_value = stats.pearsonr(df['X'], df['Y'])
print(f"Pearson's r = {r:.4f}")
print(f"p-value = {p_value:.3e}")
```

```python
slope, intercept, r_value, p_value_lr, std_err = stats.linregress(df['X'], df['Y'])
print(f"Best Price = {slope:.4f} * List Price + {intercept:.4f}")
print(f"R-squared = {r_value**2:.4f}")
```

```python
fig, ax = plt.subplots(figsize=(8, 6))

ax.scatter(df['X'], df['Y'], color='steelblue', edgecolor='black', s=60, label='Trucks (data)', zorder=3)

x_line = np.linspace(df['X'].min(), df['X'].max(), 100)
y_line = slope * x_line + intercept
ax.plot(x_line, y_line, color='firebrick', linewidth=2, label=f'Fit: y = {slope:.3f}x + {intercept:.3f}', zorder=2)

ax.set_xlabel('List Price ($1000s)', fontsize=12)
ax.set_ylabel('Best (Sale) Price ($1000s)', fontsize=12)
ax.set_title(f"Truck List Price vs. Best Price (Pearson's r = {r:.4f})", fontsize=13)
ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig('truck_scatter.png', dpi=150)
plt.show()
```

```python
list_price_choice = 20.0
predicted_best_price = slope * list_price_choice + intercept
print(f"Predicted best price for a list price of ${list_price_choice*1000:,.0f}: "
      f"${predicted_best_price*1000:,.0f}")
```

The correlation is very strong(r = 0.9985 and R² = 0.9971), so the linear relationship is very strong. This is very visible in the graph. My prediction falls on the line. I am very confident because of this.
