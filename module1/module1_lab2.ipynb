```python
import pandas as pd
df = pd.read_csv("chickwts.csv")
```

```python
df.head()
```

```python
df.groupby("feed")["weight"].agg(["min", "mean", "max"])
```

Sunflower is the correst choice. It has highest min, mean, and max. Soybean is often cheaper. Along with corn, it is also industry standard, but weight is the priority and sunflower feed produces the largest chickens across the board. Soybean ranks 3rd for min and 4th for mean and max.
