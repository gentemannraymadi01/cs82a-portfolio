```python
LB_TO_KG = 0.453592

def lb_to_kg(pounds):
    """Convert a weight in pounds to kilograms."""
    return pounds * LB_TO_KG

print(f"1 lb   = {lb_to_kg(1):.3f} kg")
print(f"10 lb  = {lb_to_kg(10):.3f} kg")

my_weight_lb = 115
print(f"My weight: {my_weight_lb} lb = {lb_to_kg(my_weight_lb):.2f} kg")
```
