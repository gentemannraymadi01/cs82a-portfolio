### migration process Colab to Kaggle

### What transfers vs. what needs redoing

| Item | Transfers as is? | What to do on Kaggle |
|---|---|---|
| .ipynb file | Yes | Import the downloaded notebook |
| Python code | Yes | Nothing |
| Markdown cells | Yes | Nothing |
| Analysis and results | Mostly | Rerun to confirm they match |
| Package installs | No | Turn on internet, then run `%pip install ...` |
| File paths | No | Change `/content/` to `/kaggle/input/<dataset-name>/...` and remove `google.colab` / `drive.mount` |
| File storage | No | Upload the data as a Kaggle dataset |
| Account | No | Create or sign in to a Kaggle account |

### Steps
1. Open the finished notebook in Colab, save and rerun it as needed
2. Download the notebook and any data files referenced from Colab or Drive
3. Create a new notebook on Kaggle and import the downloaded notebook
4. Upload the data as a Kaggle dataset
5. Delete the `google.colab` import and `drive.mount` lines
6. Change `/content/` paths to `/kaggle/input/<dataset-name>/...`
7. Turn on internet in Kaggle. It is off by default.
8. Run `%pip install ...` for any missing packages
9. Run `import pandas as pd; print(pd.__version__)` and compare to the Colab version
10. Use Run All to confirm everything works, then export from Kaggle

### one risk: package version drift
Colab and Kaggle both have separate preinstalled libraries. This means that the same notebook may run on different Python or library versions on each platform. When this happens results from the same notebook can change when it is moved and there will be no obvious warning. You can check the versions of things like pandas, numpy, and matplotlib to mitigate this.

It took me a moment to figure out that you need to upload the files to the session in Colab. I got used to uploading them to the same folder as my notebook and referencing them locally on Jupyter.
