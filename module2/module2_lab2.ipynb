```python
def multiply_for(a, b):
    """Multiply two whole numbers by repeated addition, using a for loop."""
    total = 0
    for _ in range(b):
        total = total + a
    return total
```

```python
def multiply_while(a, b):
    """Multiply two whole numbers by repeated addition, using a while loop."""
    total = 0
    count = 0
    while count < b:
        total = total + a
        count = count + 1
    return total
```

The while loop terminates when count < b, each iteration checks for that before running, so assuming b is not negative count will reach b because it starts as 0 and increases by 1 each pass.

```python
a, b = 7, 6

result_for = multiply_for(a, b)
result_while = multiply_while(a, b)
real_product = a * b

print(f"multiply_for({a}, {b})   = {result_for}")
print(f"multiply_while({a}, {b}) = {result_while}")
print(f"real product ({a} * {b}) = {real_product}")

assert result_for == real_product, "for-loop version disagrees with real product!"
assert result_while == real_product, "while-loop version disagrees with real product!"
assert result_for == result_while, "for-loop and while-loop versions disagree!"
```
