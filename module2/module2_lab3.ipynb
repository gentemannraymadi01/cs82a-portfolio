Cutoffs: Freezing < 32°F, Cold 32–49°F, Mild 50–64°F, Warm 65–84°F, Hot ≥ 85°F.

```python
temperatures = [-7, 1, 76, 7, 25, 18, 21, 104, 84, 3, 84, 70, 84, 71, 4, 59, 90, 32, 49, 50, 64, 65, 84, 85]

for temp in temperatures:
    if temp < 32:
        label = "Freezing"
    elif temp < 50:
        label = "Cold"
    elif temp < 65:
        label = "Mild"
    elif temp < 85:
        label = "Warm"
    else:
        label = "Hot"

    print(f"{temp}\u00b0F -> {label}")
```
