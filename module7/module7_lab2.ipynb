```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.read_csv("ratings.tsv", sep="\t")
print(df.shape)
df.head()
```

```python
features = ["ambience", "food", "service"]
X = df[features]
y = df["rating"]

model = LinearRegression()
model.fit(X, y)

print("Intercept:", round(model.intercept_, 4))
print("R^2:", round(model.score(X, y), 4))
print("coef_ (in column order):", model.coef_.round(4))
```

```python
ranked = (
    pd.DataFrame({"feature": features, "coefficient": model.coef_})
    .sort_values("coefficient", ascending=False)
    .reset_index(drop=True)
)
ranked.index += 1
ranked
```

Standardized-coefficient cell to check that the ranking holds

```python
std_coefs = pd.Series(
    model.coef_ * X.std().values / y.std(), index=features
).sort_values(ascending=False)
std_coefs.round(4)
```

## recommendation

Food quality is most heavily associated with score, so restaurants looking to raise their score should focus on food quality first. Every point of food score is associated with roughly 0.73 points in terms of rating holding ambience and service constant, which is more than twice the effect of service (0.32) and over five times that of ambience (0.13). This is based on only 15 restaurants and on association, so the ranking should be treated as a strong suggestion not a guarantee.
